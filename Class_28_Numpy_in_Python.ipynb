Arithmetic Operation on array


```python
# Arithmetic Operators on array perfomrs the operation elementwise

arr1 = np.random.randint(10, 20, 10)
print(arr1)
```

```python
# Mathematical Operations

print(arr1+5)
print(arr1-5)
print(arr1*5)
print(arr1/5)
print(arr1//5)
print(arr1%5)
print(arr1**5)
```

```python
arr1
```

```python
# Comparison operations
print(arr1>5)
print(arr1<5)
print(arr1<=5)
print(arr1>=5)
print(arr1==5)
print(arr1!=5)
print(arr1&5)     # It is performing bitwise operation
```

```python
# To perform these operation on the array, the dimension and shape of the array should remain same

arr1 = np.random.randint(10, 50, 5)
arr2 = np.random.randint(10, 90, 6)
print(arr1)
print(arr2)
```

```python
# Arithmetic operations - element wise
print(arr2+arr1)
print(arr2-arr1)
print(arr2*arr1)
print(arr2/arr1)
print(arr2//arr1)
print(arr2%arr1)
print(arr2**arr1)
```

```python
print(arr1 > arr2)
print(arr1 < arr2)
print(arr1 >= arr2)
print(arr1 <= arr2)
print(arr1 == arr2)
print(arr1 != arr2)
print(arr1 & arr2)
```
